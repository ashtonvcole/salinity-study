# Plot Result of Scaling Tests

```python
from matplotlib import pyplot as plt
import matplotlib.ticker
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('../tests/strong_scaling_tc/strong_scaling_tc.csv')
df['speedup'] = df['wall_s'][0] / df['wall_s']
df['efficiency'] = df['wall_s'][0] / (df['nproc'] * df['wall_s'])
```

```python
fig, ax = plt.subplots()
ax.loglog(df['nproc'], df['speedup'])
ax.loglog(df['nproc'], df['nproc'], color='k', linestyle='--')
ax.set_xlabel('cores')
ax.set_xticks(df['nproc'])
ax.xaxis.set_major_formatter(matplotlib.ticker.ScalarFormatter())
ax.xaxis.minorticks_off()
ax.set_ylabel('speedup')
ax.set_ylim((0, 50))
ax.xaxis.set_minor_formatter(matplotlib.ticker.NullFormatter())
ax.yaxis.set_major_formatter(matplotlib.ticker.ScalarFormatter())
ax.set_aspect('equal')
fig.tight_layout()
```
